```python
def calculate_area(shape:str, **kwargs):
    if shape == "square":
        area=  kwargs.get("length", 0) **2
    elif shape == "rectangle":
        if "length" not in kwargs or "breadth" not in kwargs:
            print("length and breadth are required for rectangle")
            return None
        area= kwargs.get("length", 0) * kwargs.get("breadth", 0) 
    else:
        print(f"{shape} is not supported")     
        
        return area
calculate_area("square", length=5, breadth=10) 
```

Calculate Std. Deviation of given list of numbers.
 
Standard Deviation : 
$\sqrt{\frac{\sum{(X - \bar{X})}^2}{N}}$
 

```python
def standard_deviation(numbers):
    mean = sum(numbers) / len(numbers)

    squared_differences = [(x - mean) ** 2 for x in numbers]

    variance = sum(squared_differences) / len(numbers)

    return variance ** 0.5


numbers = [10, 20, 30, 40, 50]

print("Standard Deviation:", standard_deviation(numbers))
```
