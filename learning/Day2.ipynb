Python Text Type
In Python, strings (str) are used to represent text data. Here's a brief overview of the Python string type.

Defining strings
You can define a string in Python by enclosing the text in either single or double quotes. Here are some examples:

```python
name="seiryuu"
type(name)
```

```python
number = "12345"
type(number)
```

```python
int(number)
```

```python
address = str("Kathmandu , Nepal")
address
```

String indexing
You can access individual characters in a string using indexing. In Python, indexing starts at 0. Here are some examples:

```python
address = "West Berlin, 789"
address[0]
```

```python
address[6]
```

```python
len(address)
```

```python
address[-5]
```

String slicing
You can also access a range of characters in a string using slicing. Slicing uses the syntax start:stop:step, where start is the index of the first character you want to include, stop is the index of the first character you want to exclude, and step is the number of characters to skip. Here are some examples:

```python
address = "West Berlin, 789"
# address [ start_index : stop_index : step ]
address[ 0: : 1]
```

```python
address[5:-2]
```

```python
text = "radar"
text[::-1] # Reverse
```

String Formatting
You can use string formatting to insert variables or expressions into a string. There are several ways to format strings in Python, including using % operator, .format() and f-strings. Here are some examples:

```python
name = "seiryuu"
age = 22
sent = "Your name is %s and you are %d years old"%(name.capitalize(), age)
sent
```

```python
sent = "Your name is {} and you are {} years old".format(name.capitalize(), age)
sent
```

```python
sent = f"Your name is {name.capitalize()} and you are {age} years old"
sent
```


Python Sequence Types
The sequence type of python includes two data structures: list and tuple

1. List
In Python, lists are used to store a collection of items, which can be of any data type. Here's a brief overview of the Python list type:

Defining Lists
You can define a list in Python using square brackets []. Here are some examples:

```python
prime = [1, 3, 5, 7, 11, 13]
type(prime)
```

```python
z = ["apple", 2, "-9", 0.77]
z
```

```python
len(prime)
```

Accessing List Items
You can access individual items in a list using indexing. In Python, indexing starts at 0. Here are some examples:

```python
prime = [1, 3, 5, 7, 11, 13]
prime[0]
```

```python
prime[-2]
```

Slicing Lists
You can access a range of items in a list using slicing. Slicing uses the syntax start:stop:step, where start is the index of the first item you want to include, stop is the index of the first item you want to exclude, and step is the number of items to skip. Here are some examples:

list[ start index : stop index : steps ]

```python
numbers = list(range(1,31))
print(numbers)
```

```python
odd_number = numbers[::2]
odd_number
```

```python
even_number = numbers[1::2]
even_number
```

```python
print(numbers[::-1])
```

Modifying Lists
You can modify individual items in a list by assigning a new value to the index. You can also add or remove items from a list using built-in list methods. Here are some examples:

Docs : https://python-reference.readthedocs.io/en/latest/docs/list/

```python
fruits = ["apple", "oranges"]
fruits
```

```python
fruits.append("kiwi") # insert at last index
fruits
```

```python
fruits.insert(1, "banana") # insert at index 1
fruits
```

```python
fruits.extend(["mango", "grapes"])
fruits
```

```python
fruits[2] = "pineapple"
fruits
```

```python
fruits.remove("kiwi")
fruits
```

```python
fruits.pop(3) # Delete uing list index
fruits
```
