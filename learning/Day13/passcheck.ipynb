```python
class PassCheck:
    def __init__(self, password:str):
        self.password = password

    def check_password(self):
        if len(self.password) >= 8 and any(c.isalpha() for c in self.password) and any(c.isdigit() for c in self.password):
    
            return "Strong Password"
        else:
            return "Weak Password"
password = input("Enter password: ")

p = PassCheck(password)

print(p.check_password())
```
