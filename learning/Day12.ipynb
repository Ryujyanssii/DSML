Calculating Student Results
Develop a class to accept a student's name and marks in three subjects, then calculate and display the total and average marks.
 
total() , average() , display()
 

```python
class Student:
    def __init__(self, name, mark1, mark2, mark3):
        self.name = name
        self.mark1 = mark1
        self.mark2 = mark2
        self.mark3 = mark3

    def total(self):
        return self.mark1 + self.mark2 + self.mark3

    def average(self):
        return self.total() / 3

    def display(self):
        print("Student Name:", self.name)
        print("Total Marks:", self.total())
        print("Average Marks:", self.average())


# Create object
student1 = Student("Rujan", 80, 75, 90)

# Display result
student1.display()
```

Designing a simple hierarchy of employees in a company
 
    - The Person class stores basic details like name and age.
 
    - The Employee class inherits from Person and adds an employee ID.
 
    - The Manager class inherits from Employee and adds a department.
 
Write these classes and implement a method get_details() in the Manager class that returns a string with name, age, employee ID, and department.
 

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age


class Employee(Person):
    def __init__(self, name, age, employee_id):
        super().__init__(name, age)
        self.employee_id = employee_id


class Manager(Employee):
    def __init__(self, name, age, employee_id, department):
        super().__init__(name, age, employee_id)
        self.department = department

    def get_details(self):
        return f"Name: {self.name}, Age: {self.age}, Employee ID: {self.employee_id}, Department: {self.department}"


# Create Manager object
manager = Manager("Rujan", 22, "EMP101", "IT")

print(manager.get_details())
```
