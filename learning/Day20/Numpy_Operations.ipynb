```python
a = np.array(
    [
        [4,6,8], 
        [1,2,4]
    ]
)
b = np.random.randint(1, 45, size = (2,3))
c = np.linspace((0,5), (60,100), num = 4, dtype=int)
a, b, c
```

```python
# a + b
np.add(a,b)
```

```python
# a ** 2
np.pow(a, 2)
```

2. Matrix Operations

```python
# Transpose

# b.T
bT = np.transpose(b)
bT
```

```python
# Matrix Multiplication

# a @ bT
# np.dot(a, bT)
res = a.dot(bT)
res
```

```python
# Trace [Diagnoal sum]
d_sum = np.trace(res)
print(d_sum)
```

```python
# Rank
np.linalg.matrix_rank(a)
```

```python
# Determinat
b = np.random.randint(1, 5, size = (3,3))

det_b = np.linalg.det(b)
det_b
```

```python
# Inverse
inv_b = np.linalg.inv(b)
inv_b
```

Solve Equation : 4x + 2y = 8 , 5x + 3y = 9
