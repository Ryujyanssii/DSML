Data Encoding
Data Encoding is a technique that is used to convert categorical columns into numerical ones so that they can be fitted by machine learning models which only take numerical data. It is an important pre-processing step in a machine-learning project.


```python
### Nominal Data  -> data without natural order
titanic_df.Sex
```

```python
# Sklean

from sklearn.preprocessing import OrdinalEncoder, LabelEncoder
ordinal_encoder = OrdinalEncoder(dtype=int)
encoded_result = ordinal_encoder.fit_transform(titanic_df[["Embarked"]])
encoded_result
```

```python
ordinal_encoder.categories_
```

```python
ordinal_encoder.inverse_transform(
    [[1],[2],[0],[1]]
)
```

```python
final_df = titanic_df.drop(
    ["Name", "Ticket", "initials", "encoded_embarked"], axis=1
)
onehot_encoder = OneHotEncoder(sparse_output=False)
ordinal_encoder = OrdinalEncoder(dtype = int)
age_embarked_encoded = onehot_encoder.fit_transform(titanic_df[["Sex", "Embarked"]])
pclass_encoded = ordinal_encoder.fit_transform(final_df[["Pclass"]])
```

```python
final_df.Pclass = pclass_encoded
final_df = pd.get_dummies(final_df, columns=["Sex", "Embarked"], dtype=int)
final_df.head()
```

```python
final_df.to_csv("../encodedtitanic_dataset.csv", index = False)
```
